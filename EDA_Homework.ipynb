### Data Cleaning Decisions

The dataset originally contained 891 rows and 15 columns.

- Missing values in `embarked` and `embark_town` were replaced with the most frequent value (mode).
- Missing `age` values were replaced using the median age within each sex group.
- The `deck` column was removed because approximately 77.22% of its values were missing.
- No rows were removed during the cleaning process.

After cleaning, the dataset contains 891 rows and 14 columns.

### Duplicate Rows

The dataset contains 107 duplicate rows, representing approximately 12.01% of the dataset.

These duplicate rows were retained because the dataset does not contain a unique passenger identifier. Therefore, identical rows cannot be assumed to represent the same passenger. Removing them could incorrectly eliminate valid observations.

```python
survival_by_sex = df.groupby("sex")["survived"].mean() * 100
survival_by_sex
```

### Interpretation

The survival rate was substantially different between the two sexes. Approximately 74.2% of female passengers survived, compared with approximately 18.9% of male passengers. This indicates a strong association between sex and survival in this dataset. However, this association does not by itself establish causation.

```python
sns.barplot(x=survival_by_sex.index, y=survival_by_sex.values)

plt.title("Survival Rate by Sex")
plt.xlabel("Sex")
plt.ylabel("Survival Rate (%)")
plt.ylim(0, 100)
plt.show()
```

### Interpretation

The bar plot shows a substantial difference in survival rates between female and male passengers. Approximately 74.2% of female passengers survived, compared with 18.9% of male passengers.

```python
survival_by_class = df.groupby("pclass")["survived"].mean() * 100
survival_by_class
```

### Interpretation

The survival rate varied across passenger classes. Approximately 63.0% of first-class passengers survived, compared with 47.3% of second-class passengers and 24.4% of third-class passengers. The results show an association between passenger class and survival in this dataset.

```python
sns.barplot(x=survival_by_class.index, y=survival_by_class.values)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate (%)")
plt.ylim(0, 100)
plt.show()
```

### Interpretation

The bar plot shows that survival rates varied across passenger classes. Approximately 63.0% of first-class passengers survived, compared with 47.3% of second-class passengers and 24.4% of third-class passengers.

```python
numeric_cols = ["survived", "pclass", "age", "sibsp", "parch", "fare"]

corr = df[numeric_cols].corr()

corr
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap of Numerical Variables")
plt.show()
```

### Interpretation

The correlation heatmap shows several relationships among the numerical variables. The strongest observed relationship is the negative correlation between passenger class and fare (-0.55). A moderate positive correlation is observed between sibsp and parch (0.41). Survival has a negative correlation with pclass (-0.34) and a positive correlation with fare (0.26). The correlation between survival and age is weak (-0.07). Correlation indicates association and does not establish causation.

```python
sns.scatterplot(
    data=df,
    x="age",
    y="fare",
    hue="survived",
    alpha=0.7
)

plt.title("Age vs Fare by Survival Status")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

### Interpretation

The scatter plot shows the relationship between passenger age and fare, with points distinguished by survival status. Most observations are concentrated at lower fare values, while a small number of passengers paid much higher fares. The survival groups overlap considerably. The correlation between age and fare is approximately 0.09, indicating a very weak linear relationship between the two variables.

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="age", bins=20, kde=True)

plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```

### 1. Main Characteristics of the Titanic Dataset

The Titanic dataset originally contained 891 rows and 15 columns. After data cleaning, it contains 891 rows and 14 columns because the `deck` column was removed due to a high percentage of missing values. The dataset contained missing values and 107 duplicate rows. The number of male passengers was higher than the number of female passengers, while the survival rate was much higher among females. The survival rate was also higher for first-class passengers than for second-class and third-class passengers.

### 2. Missing Values and Duplicate Rows

The dataset contained missing values in four columns. The `age` column had 177 missing values, which were replaced using the median age calculated separately for each sex. The `embarked` and `embark_town` columns each had 2 missing values, which were replaced using the mode.

The `deck` column contained 688 missing values (approximately 77.22% of the dataset). Because of this high percentage of missing data, the column was removed from the dataset.

The dataset also contained 107 duplicate rows. These rows were retained because the dataset does not contain a unique passenger identifier, so identical rows cannot be assumed to represent the same passenger.

### 3. Distribution of Passengers by Sex and Passenger Class

The dataset contains more male passengers than female passengers. There are 577 male passengers and 314 female passengers. Regarding passenger class, the third class contains the largest number of passengers, followed by the first and second classes. However, the survival rate was higher among first-class passengers (62.96%) than second-class passengers (47.28%) and third-class passengers (24.36%).

### 4. Survival Rate by Sex

The survival rate among female passengers was much higher than the survival rate among male passengers. Approximately 74.2% of female passengers survived, compared with 18.9% of male passengers.

### 5. Survival Rate by Passenger Class

The survival rate was higher among first-class passengers than second-class passengers, and higher among second-class passengers than third-class passengers. The survival rates were 62.96%, 47.28%, and 24.36%, respectively.

### 6. Relationships Between Numerical Variables

The correlation analysis shows a noticeable negative relationship between `pclass` and `fare` (-0.55). There is a moderate positive relationship between `sibsp` and `parch` (0.41). The variable `survived` has a weak-to-moderate negative relationship with `pclass` (-0.34) and a weak positive relationship with `fare` (0.26). Correlation indicates association and does not establish causation.

## F. Conclusions

The exploratory data analysis of the Titanic dataset revealed several important patterns. The dataset originally contained 891 rows and 15 columns and was reduced to 14 columns after removing the `deck` column because of its high proportion of missing values. Missing values in `age`, `embarked`, and `embark_town` were handled using median and mode imputation.

The analysis showed that there were more male passengers than female passengers, but the survival rate was substantially higher among female passengers. Survival rates also varied by passenger class, with the highest rate in first class and the lowest in third class.

Correlation analysis identified several relationships among numerical variables, including a noticeable negative relationship between `pclass` and `fare` and a moderate positive relationship between `sibsp` and `parch`. These relationships represent associations in the dataset and should not be interpreted as evidence of causation.

One limitation is that the dataset contains duplicate rows and does not provide a unique passenger identifier, so the duplicate rows were retained rather than removed.
